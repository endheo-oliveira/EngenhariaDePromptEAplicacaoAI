```python
#Calculadora simples

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
#Pergunta ao operação matematica o usuario quer
operacao = input("Digite a operação (+, -, *, /): ")

#Busca opersador digitado
operadores = {
    '+': lambda a, b: a + b,
    '-': lambda a, b: a - b,
    '*': lambda a, b: a * b,
    '/': lambda a, b: a / b,
}

#Testa se o operador esta no dicionario operadores
if operacao in operadores:
    #Se operador tiver no dicionario operadores vai executar o calculo
    num = operadores[operacao](num1, num2)
    print("O resultado da operação:", num)
#Se não o codigo encerra
else:
    print("Operação inválida")
```
